```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
house_price = pd.read_csv("train (1).csv")
```

```python
house_price.head()
```

```python
plt.hist(house_price["SalePrice"].dropna(), bins=30, edgecolor="black")

plt.xlabel("Sale Price ($)")
plt.ylabel("Number of Houses")
plt.title("Distribution of House Sale Prices")
```

```python
house_price["SalePrice"].describe()
```

```python
house_price["SalePrice"].median()
```

```python
mssubclass = house_price['MSSubClass'].value_counts().sort_index()
```

```python
mssubclass.plot(kind ="bar")
plt.xlabel('House Classes')
plt.ylabel('Number of Houses')
plt.title('Distribution of Subclass')
```

```python
house_price.boxplot(column="SalePrice", by="MSSubClass")

plt.suptitle("")
plt.xlabel("House Subclass")
plt.ylabel("Sale Price ($)")
plt.title("Sale Price Distribution by House Subclass")
```

```python
plt.scatter(house_price["LotArea"], house_price["SalePrice"])
plt.xlabel("Lot Area (sq ft)")
plt.ylabel("Sale Price ($)")
plt.title("Lot Area vs. Sale Price")
```

```python
correlation1 = house_price["LotArea"].corr(house_price["SalePrice"])
print("Correlation:", correlation1)
```

```python
plt.scatter(house_price['GrLivArea'], house_price['SalePrice'])
plt.xlabel("Above-Ground Living Area (sq ft)")
plt.ylabel("Sale Price ($)")
plt.title("Living Area vs. Sale Price")
```

```python
correlation2 = house_price["GrLivArea"].corr(house_price["SalePrice"])
print("Correlation:", correlation2)
```

```python
house_price.boxplot(column="SalePrice", by="OverallQual")
plt.suptitle("Overall Quality vs. Sales Price")
plt.xlabel("Overall Quality")
plt.ylabel("Sale Price ($)")
```

```python
house_price.boxplot(column="SalePrice", by="BedroomAbvGr")
plt.suptitle("Number of bedrooms vs Sale Price")
plt.xlabel("Number of Bedrooms")
plt.ylabel("Sale Price ($)")
```
