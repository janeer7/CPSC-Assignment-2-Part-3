```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
house_price = pd.read_csv("train (1).csv")
```

```python
house_price.head()
```

```python
plt.hist(house_price["SalePrice"].dropna(), bins=30, edgecolor="black")

plt.xlabel("Sale Price ($)")
plt.ylabel("Number of Houses")
plt.title("Distribution of House Sale Prices")
```

```python
house_price["SalePrice"].describe()
```

```python
house_price["SalePrice"].median()
```

```python
numeric_data = house_price.select_dtypes(include=['number'])
saleprice_corr = numeric_data.corr()['SalePrice'].sort_values(ascending=False)
correlation_table = saleprice_corr.reset_index()
correlation_table.columns = ['Variable', 'Correlation with SalePrice']
display(correlation_table)
```

```python
correlation_table = saleprice_corr.drop('SalePrice')
correlation_table.sort_values().plot(kind='barh', figsize = (10,12))

plt.xlabel('Correlation')
plt.ylabel('Variables')
plt.title('Correlation of Numerical Variables with SalePrice')
plt.tight_layout()
plt.savefig("Correlation with Sales Price")
```

```python
house_price.boxplot(column="SalePrice", by="OverallQual")
plt.suptitle("Overall Quality vs. Sales Price")
plt.xlabel("Overall Quality")
plt.ylabel("Sale Price ($)")
```

```python
plt.scatter(house_price['GrLivArea'], house_price['SalePrice'])
plt.xlabel("Above-Ground Living Area (sq ft)")
plt.ylabel("Sale Price ($)")
plt.title("Living Area vs. Sale Price")
```
